# Suppressing dust emission in 17517

```python
import prospect.io.read_results as reader
from prospector_utils.params import get_MAP
from prospect.sources import FastStepBasis
from prospect.utils.plotting import posterior_samples

plot_dir = '/Users/benjamincollins/University/master/Red_Cardinal/prospector/appphot_only_wMIRI/fits_nodust/'
stats_dir = '/Users/benjamincollins/University/master/Red_Cardinal/prospector/appphot_only_wMIRI/pickle_files_nodust/'
phot_table = '/Users/benjamincollins/University/Master/Red_Cardinal/photometry/phot_tables/fits/Phot_Table_MIRI.fits'

# Load the h5 file for the given objid
data_dir = '/Users/benjamincollins/Data/Bluejay/Prospector/v2.0.5/appphot_only_wMIRI_cat/'

objid = 17517

h5_file = glob.glob(os.path.join(data_dir, f"output_{objid}*.h5"))

try:
    h5_file = h5_file[0]
except IndexError:
    print(f"No PROSPECTOR results found for objid {objid}.")

# Load PROSPECTOR results
full_path = os.path.join(data_dir, h5_file)
results, obs, model = reader.results_from(full_path)

filename = os.path.join(stats_dir, f"{objid}.pkl")

#print(obs['filters'])
#return

# Now we have to exclude the last 3 parameters from the fit
map_parameters = get_MAP(results)

#print(map_parameters)

# Build the MAP dictionary
MAP = {}
for a,b in zip(results['theta_labels'], map_parameters):
    MAP[a] = b
    
zred = MAP['zred']
logmass = MAP['logmass']
dust2 = MAP['dust2']    # extract the diffuse dust V-band optical depth

# Calculate the spectrum based on the Maximum A Posteriori (MAP) parameters
sps = FastStepBasis(zcontinuous=1)

# 1. Create a copy of your MAP parameters
no_dust_params = map_parameters.copy()

# 2. Toggle the model setting to False
# This prevents the code from adding the IR 'glow'
model.params['add_dust_emission'] = np.array([False])

# 3. Predict the spectrum
# The resulting spec/phot will show the attenuated UV but NO IR emission
spec, phot, _ = model.predict(no_dust_params, obs=obs, sps=sps)    

# Convert maggies to µJy
maggies_to_muJy = 3631e6

# Wavelengths of the model spectrum
wave_spec = sps.wavelengths

# Convert to arrays
phot = np.array(phot)

# Draw 100 posterior samples
samples = posterior_samples(results, 100)

sample_specs = []
for params_i in samples:    
    spec_i, _, _ = model.predict(params_i, obs=obs, sps=sps)
    sample_specs.append(spec_i)
sample_specs = np.array(sample_specs)  # shape: (nsample, nwave)

# Takes the per-pixel percentiles such that the final spectra are not actual spectra of Prospectors parameter space
lower = np.percentile(sample_specs, 16, axis=0)
median = np.percentile(sample_specs, 50, axis=0)
upper = np.percentile(sample_specs, 84, axis=0)

# Compute filter wavelength in microns
phot_wave = np.array([filt.wave_effective for filt in obs['filters']])  # in Angstroms

data = {
    # Important metadata
    'id': objid,
    'zred': zred,
    'maggies_to_muJy': maggies_to_muJy,
    
    'model': {
        'spec_best': spec,
        'spec_16th': lower,
        'spec_median': median,
        'spec_84th': upper,
        'wave_spec': wave_spec,
        'sample_specs': sample_specs[:10],
        'phot': phot,
        'phot_wave': phot_wave
    },
    
    # One entry for the observation dictionary
    'obs': obs,
    
    #'fit_quality': fit_quality,
    
    'galaxy_properties': {
        'logmass': logmass,
        'dust2': dust2
    },
    
    # Moved it outside so it's easier accessible
    'map_theta': MAP # Keep the full raw dictionary just in case    
}

# Write output to a pickle file
with open(filename, 'wb') as f:
    pkl.dump(data, f)
print(f"💾 Saved data to {filename}")
    
if plot_dir:
    plot_miri_fit(filename, plot_dir)
```

Restructure pickle files to have map_parameters easily accessible:

```python
stats_dir = '/Users/benjamincollins/University/master/Red_Cardinal/prospector/sourcephotonly/pickle_files'

output_folder = '/Users/benjamincollins/University/master/Red_Cardinal/prospector/sourcephotonly/pickle_files_new'

os.makedirs(output_folder, exist_ok=True)

file_paths = glob.glob(os.path.join(stats_dir, "*.pkl"))
print(f"Found {len(file_paths)} files. Starting restructuring...")

for f_path in file_paths:
    with open(f_path, 'rb') as f:
        old_data = pkl.load(f)

    # 1. Extract raw MAP dictionary
    gal_props = old_data.get('galaxy_properties', {})
    
    MAP = gal_props.get('map_theta', {})
    
    # 2. Extract observational data
    # If obs_miri isn't a separate key yet, we take it from 'obs'
    obs_data = old_data.get('obs', {})

    # 3. Reconstruct Galaxy Properties
    # We pull these from the existing map_theta
    galaxy_props = {
        'logmass': gal_props.get('logmass'),
        'dust2': gal_props.get('dust2'),
        'sfr_100myr': gal_props.get('sfr_100myr'), # Call helper for math
        'sfr_30myr': gal_props.get('srf_30myr'), 
        'sfr_bins': gal_props.get('sfr_bins'), # Add your specific bin array if available
        'agebins': gal_props.get('agebins')   # Add your specific agebins array if available
    }

    # 4. Create New Structure
    new_data = {
        'id': old_data.get('id'),
        'zred': old_data.get('zred'),
        'maggies_to_muJy': old_data.get('maggies_to_muJy'),
        'model': old_data.get('model'),
        'obs': obs_data,
        'fit_quality': old_data.get('fit_quality', {}), # Carry over if exists
        'galaxy_properties': galaxy_props,
        'map_theta': MAP # Now outside and easily accessible
    }

    # 5. Save the new file
    new_filename = os.path.join(output_folder, os.path.basename(f_path))
    with open(new_filename, 'wb') as f:
        pkl.dump(new_data, f)

print("Restructuring complete.")
```

```python
pickle_dir = '/Users/benjamincollins/University/master/Red_Cardinal/prospector/appphot_only_wMIRI/flux_scaled/pickle_files/'
plot_dir = '/Users/benjamincollins/University/master/Red_Cardinal/prospector/appphot_only_wMIRI/flux_scaled/fit_quality/'
os.makedirs(plot_dir, exist_ok=True)

pickle_files = glob.glob(f'{pickle_dir}/*.pkl')

all_data = []
reduced_chi2_list = []

# 1. DATA EXTRACTION
for filename in pickle_files:
    with open(filename, 'rb') as f:
        data = pkl.load(f)
        
    gid = data['id']
    
    if gid in exclude:
        continue
    
    fit_quality = data.get('fit_quality', {})
    
    # Store global per-galaxy stats
    if 'chi2_red' in fit_quality:
        reduced_chi2_list.append({
            'galaxy_id': gid,
            'reduced_chi2': fit_quality['chi2_red'],
            'n_filters': len(fit_quality) - 1 # Assuming only 'chi2_red' is non-filter
        })

    # Store per-band stats
    for i, (key, val) in enumerate(fit_quality.items()):
        if isinstance(val, dict): # This identifies the filter entries
            all_data.append({
                'galaxy_id': gid,
                'filter_name': key,
                'N_sigma': val.get('n_sigma') * (-1),
                'flux': val.get('obs_flux'),
                'flux_err': val.get('obs_err'),
                'flux_mod': val.get('mod_flux'),
                'flux_mod_err': val.get('mod_err'),
                'frac_diff': val.get('frac_diff') * (-1),
                'snr': val.get('snr')
            })
            
            if val.get('n_sigma') > 7.0:
                print(gid, key)
                print(np.log10(val.get('obs_flux')))
                print(np.log10(val.get('mod_flux')))
                print("Ratio:", np.log10(val.get('obs_flux')/val.get('mod_flux')))
                print("Nsigma:", val.get('n_sigma'))
                print("\n")

df = pd.DataFrame(all_data)
chi2_df = pd.DataFrame(reduced_chi2_list)


bands = ['F770W', 'F1000W', 'F1800W', 'F2100W']
colors = ['#1f77b4', '#2ca02c', '#ff7f0e', '#d62728']  # Distinct colors per band
```

Plot the log-ratios

```python
# PLOT 2: LOG RATIOS
fig, axes = plt.subplots(2, 2, figsize=(10, 8), sharex=False, sharey=False)
axes = axes.flatten()  # easier to index

for i, (ax, band) in enumerate(zip(axes, bands)):
    subset = df[df['filter_name'] == band]
    
    f_obs = subset['flux']
    err = subset['flux_err']
    frac_diff = subset['frac_diff']
    
    # SNR filter: Keep only detections > 3-sigma
    snr_mask = (f_obs / err) >= 3.0
    
    # Model reconstruction
    # Using the simplified: f_model = f_obs * (1 - frac_diff)
    # We filter out frac_diff >= 1 to avoid log(0) or log(negative)
    valid_mask = (frac_diff < 1.0) & snr_mask
    
    # Calculate ratios only for valid entries
    # log10(f_obs / (f_obs * (1-frac_diff))) reduces to -log10(1-frac_diff)
    log_ratios = -np.log10(1.0 - frac_diff[valid_mask])
    
    ax.set_title(f'{band}')
    #ax.set_xlim(x_min, x_max)
    ax.set_xlabel('Flux ratio (dex)')
    ax.set_ylabel('Number of galaxies')
    
    #if i in [0,1]: ax.set_ylim(0, 24)
    #elif i in [2,3]: ax.set_ylim(0,12)
    
    # Add compact statistics
    mean_logr = np.mean(log_ratios)
    std_logr = np.std(log_ratios)
    median_logr = np.median(log_ratios)
    mad_logr = median_abs_deviation(log_ratios)
    N = len(log_ratios)
    num = f'N = {N}'
    
    x_min = -1.2
    x_max = 1.2
    bins = np.linspace(x_min, x_max, 25)
    
    counts, _, _ = ax.hist(log_ratios, bins=bins, color=colors[i], alpha=0.7, edgecolor='black')

    ymax = np.max(counts) * 1.1 # for all plots
    ymax = max(ymax, 10)
    ax.set_ylim(0,ymax)
    ax.vlines(median_logr, ymin=0, ymax=ymax, color='darkred', alpha=0.8, linestyle='-', linewidth=2, label=f'Median: {median_logr:.2f}')
#            if i == 2: 
#               stats_text += ' (*)'
#              print(log_ratios[log_ratios > 1])
    ax.text(0.025, 0.92, num, transform=ax.transAxes, fontsize=10,
            bbox=dict(boxstyle='round,pad=0.3', facecolor='white', alpha=0.8))
    ax.plot([],[], label=f'MAD: {mad_logr:.2f}', alpha=0)  # dummy plot for legend
    ax.legend()
    
#plt.suptitle(r'$N_\sigma$ distribution for each MIRI filter', fontsize=14)
plt.tight_layout(rect=[0, 0, 1, 0.96])
# Save single combined figure
filename = os.path.join(plot_dir, 'log_ratios.png')
plt.savefig(filename, dpi=300)
plt.show()
```

```python
import os
import matplotlib.pyplot as plt
import matplotlib.ticker as ticker
from scipy.stats import binned_statistic
import numpy as np
import pandas as pd

# Load results table
df_lum = pd.read_csv(os.path.join(out_dir, 'LIR_results.csv'))

# ----------------------------------------------------------------------
# 1. Parse Chain Strings to NumPy Float Arrays
# ----------------------------------------------------------------------
def parse_chain_string(chain_str):
    """Converts a string representation of a numpy array into a float numpy array."""
    if isinstance(chain_str, np.ndarray):
        return chain_str
    clean = str(chain_str).strip("'\"[] \n").replace('\n', ' ')
    return np.fromstring(clean, sep=' ', dtype=float)

df_lum['lir_chain_miri'] = df_lum['lir_chain_miri'].apply(parse_chain_string)
df_lum['lir_chain_nomiri'] = df_lum['lir_chain_nomiri'].apply(parse_chain_string)

# ----------------------------------------------------------------------
# 2. Compute Per-Galaxy Ratio Distributions and 16th/50th/84th Percentiles
# ----------------------------------------------------------------------
def compute_galaxy_ratio_stats(row):
    c_m = row['lir_chain_miri']
    c_nom = row['lir_chain_nomiri']
    
    # Outer difference (100, 1) - (1, 100) -> 10,000 draws of delta = log(MIRI) - log(noMIRI)
    delta_samples = (c_m[:, None] - c_nom[None, :]).ravel()
    ratio_samples = 10.0**delta_samples
    
    med = np.nanmedian(ratio_samples)
    p16 = np.nanpercentile(ratio_samples, 16)
    p84 = np.nanpercentile(ratio_samples, 84)
    
    return pd.Series({
        'ratio_med': med,
        'err_low': med - p16,
        'err_high': p84 - med
    })

stats_df = df_lum.apply(compute_galaxy_ratio_stats, axis=1)
df_lum = pd.concat([df_lum, stats_df], axis=1)

# Extract arrays for plotting
dust = df_lum['dust_nomiri'].values
ratio = df_lum['ratio_med'].values

# Format asymmetric errors for matplotlib: shape (2, N)
yerr = np.array([
    df_lum['err_low'].values,
    df_lum['err_high'].values
])

# Optional: Horizontal error bars for Av if present in df_lum
xerr = None
if 'dust_lo_nomiri' in df_lum and 'dust_hi_nomiri' in df_lum:
    xerr = np.array([
        df_lum['dust_lo_nomiri'].values,
        df_lum['dust_hi_nomiri'].values,
    ])

# ----------------------------------------------------------------------
# 3. Running Binned Statistics
# ----------------------------------------------------------------------
nbins = 5
bin_edges = np.linspace(np.nanmin(dust), np.nanmax(dust), nbins + 1)
bin_centers = 0.5 * (bin_edges[:-1] + bin_edges[1:])

medians, _, _ = binned_statistic(dust, ratio, statistic='median', bins=bin_edges)
p16, _, _     = binned_statistic(dust, ratio, statistic=lambda x: np.nanpercentile(x, 16), bins=bin_edges)
p84, _, _     = binned_statistic(dust, ratio, statistic=lambda x: np.nanpercentile(x, 84), bins=bin_edges)

# ----------------------------------------------------------------------
# 4. Plot Setup
# ----------------------------------------------------------------------
fig, ax = plt.subplots(figsize=(4, 3.6))  # Single-column A&A format

# 1:1 Unity Line
ax.axhline(y=1.0, color='grey', linestyle='--', linewidth=1.5, zorder=1, label='Unity')

# Binned running trend and 1-sigma uncertainty band
ax.plot(bin_centers, medians, color='crimson', lw=1.5, label='Running median', zorder=3, alpha=0.85)
ax.fill_between(bin_centers, p16, p84, color='crimson', alpha=0.15,
                label=r'$1\sigma$ uncertainty', zorder=2)

# Error bars (drawn under scatter points)
ax.errorbar(
    dust,
    ratio,
    xerr=xerr,
    yerr=yerr,
    fmt='none',
    ecolor='gray',
    elinewidth=0.6,
    capsize=1.5,
    capthick=0.6,
    alpha=0.45,
    zorder=4,
)

# Scatter points on top
ax.scatter(
    dust,
    ratio,
    s=32,
    facecolor='dodgerblue',
    edgecolor='black',
    linewidth=0.4,
    alpha=0.75,
    zorder=5
)

# Scales and limits
ax.set_yscale('log')
ax.set_xlim(-0.1, 3.2)
ax.set_ylim(0.08, 15.0)

ax.set_xlabel(r'$A_V$', fontsize=12)
ax.set_ylabel(r'$L_{\mathrm{IR}}^{\mathrm{MIRI}} \,/\, L_{\mathrm{IR}}^{\mathrm{no\,MIRI}}$', fontsize=12)
ax.tick_params(labelsize=11, direction='in', top=True, right=True, which='both')

ax.legend(loc='upper right', fontsize=8.5, framealpha=0.9)
plt.tight_layout()

out_plot_png = os.path.join(out_dir, 'LIR_log_ratios_errs_improved.png')
plt.savefig(out_plot_png, dpi=300, bbox_inches='tight')
print(f"✅ Figure saved to: {out_plot_png}")
plt.show()

print("Binned Medians:", medians)
```

```python
import os
import glob
import pickle as pkl
import numpy as np
import pandas as pd
import scipy.stats as stats
import matplotlib.pyplot as plt

# ----------------------------------------------------------------------
# 1. Directories & Parameter Setup
# ----------------------------------------------------------------------
sourcephotonly = '/Users/benjamincollins/University/Master/Red_Cardinal/prospector/sourcephotonly/flux_scaled/pickle_files_v2'
appphot_only_wMIRI = '/Users/benjamincollins/University/Master/Red_Cardinal/prospector/appphot_only_wMIRI/flux_scaled/pickle_files_v2'
out_plot_dir = '/Users/benjamincollins/University/Master/Red_Cardinal/prospector/miri_comparison/'
os.makedirs(out_plot_dir, exist_ok=True)

labels = [
    'logzsol', 'logmass', 'dust2', 'dust_index', 'dust1_fraction',
    'gas_logz', 'gas_logu', 'duste_umin', 'duste_qpah', 'duste_gamma'
]

# Formatted LaTeX titles for publication
tex_labels = {
    'logzsol': r'Stellar Metallicity $\log(Z/Z_\odot)$',
    'logmass': r'Stellar Mass $\log(M_*/M_\odot)$',
    'dust2': r'Diffuse Dust Optical Depth $\tau_{V,2}$',
    'dust_index': r'Attenuation Curve Slope $n$',
    'dust1_fraction': r'Birth Cloud Fraction ($Q_{\mathrm{dust1}}$)',
    'duste_umin': r'Min Radiation Field $U_{\mathrm{min}}$',
    'duste_qpah': r'PAH Mass Fraction $q_{\mathrm{PAH}}$ [%]',
    'duste_gamma': r'PDR Dust Fraction $\gamma_e$',
    'gas_logz': r'Gas Metallicity $\log(Z_{\mathrm{gas}}/Z_\odot)$',
    'gas_logu': r'Ionisation Parameter $\log(U)$'
}

# ----------------------------------------------------------------------
# 2. Extraction Function
# ----------------------------------------------------------------------
def extract_single_param(pdict):
    """Safely extracts median, err_low, err_high from a parameter entry."""
    if not isinstance(pdict, dict):
        return np.nan, np.nan, np.nan
    
    val = pdict.get('median', pdict.get('map', np.nan))
    err_lo = pdict.get('err_low', np.nan)
    err_hi = pdict.get('err_high', np.nan)
    
    # Fallback if err_low/err_high are missing but p16/p84 exist
    if np.isnan(err_lo) and 'p16' in pdict:
        err_lo = np.maximum(0.0, val - pdict['p16'])
    if np.isnan(err_hi) and 'p84' in pdict:
        err_hi = np.maximum(0.0, pdict['p84'] - val)
        
    return val, err_lo, err_hi

records = []

for objid in valid_ids:
    if 'ids_to_exclude' in locals() and objid in ids_to_exclude:
        continue

    f_no = glob.glob(os.path.join(sourcephotonly, f"*{objid}*.pkl"))
    f_miri = glob.glob(os.path.join(appphot_only_wMIRI, f"*{objid}*.pkl"))
    
    if len(f_no) == 0 or len(f_miri) == 0:
        continue
        
    try:
        with open(f_no[0], 'rb') as p:
            d_no = pkl.load(p)
        with open(f_miri[0], 'rb') as p:
            d_miri = pkl.load(p)
    except Exception as e:
        print(f"⚠️ Error reading objid {objid}: {e}")
        continue

    p_no = d_no.get('fit_params', {})
    p_miri = d_miri.get('fit_params', {})

    entry = {'id': objid}
    for p_name in labels:
        val_no, elo_no, ehi_no = extract_single_param(p_no.get(p_name, {}))
        val_mi, elo_mi, ehi_mi = extract_single_param(p_miri.get(p_name, {}))

        entry[f'{p_name}_no'] = val_no
        entry[f'{p_name}_no_elo'] = elo_no
        entry[f'{p_name}_no_ehi'] = ehi_no

        entry[f'{p_name}_miri'] = val_mi
        entry[f'{p_name}_miri_elo'] = elo_mi
        entry[f'{p_name}_miri_ehi'] = ehi_mi

    records.append(entry)

df_params = pd.DataFrame(records)
print(f"Successfully extracted {len(df_params)} matching galaxies.")

# ----------------------------------------------------------------------
# 3. Plotting: 2x5 Grid of Parameter Comparisons
# ----------------------------------------------------------------------
fig, axes = plt.subplots(2, 5, figsize=(20, 8))
axes = axes.flatten()

for i, p_name in enumerate(labels):
    ax = axes[i]
    
    if p_name in ['logzsol', 'logmass', 'gas_logz', 'gas_logu']:
            colour = 'dodgerblue'
    elif p_name in ['dust2', 'dust_index', 'dust1_fraction']:
        colour = 'indianred'
    elif p_name in ['duste_umin', 'duste_gamma', 'duste_qpah']:
        colour = 'darkorange'
    
    # X: With MIRI, Y: Without MIRI
    x = df_params[f'{p_name}_miri'].values
    y = df_params[f'{p_name}_no'].values
    
    x_err = np.array([
        df_params[f'{p_name}_miri_elo'].values,
        df_params[f'{p_name}_miri_ehi'].values
    ])
    y_err = np.array([
        df_params[f'{p_name}_no_elo'].values,
        df_params[f'{p_name}_no_ehi'].values
    ])
    
    # Draw error bars underneath
    ax.errorbar(
        x, y,
        xerr=x_err,
        yerr=y_err,
        fmt='none',
        ecolor='gray',
        elinewidth=0.7,
        capsize=1.5,
        capthick=0.7,
        alpha=0.35,
        zorder=1
    )
    
    # Scatter points
    ax.scatter(
        x, y,
        s=40,
        color=colour,
        edgecolor='black',
        linewidth=0.5,
        alpha=0.8,
        zorder=3
    )
    
    # Dynamic axis limits & 1:1 line
    valid = np.isfinite(x) & np.isfinite(y)
    if np.sum(valid) > 2:
        all_vals = np.concatenate([x[valid], y[valid]])
        v_min, v_max = np.nanmin(all_vals), np.nanmax(all_vals)
        pad = 0.08 * (v_max - v_min) if v_max > v_min else 0.2
        lims = [v_min - pad, v_max + pad]
        
        ax.plot(lims, lims, 'k--', lw=1.2, alpha=0.6, zorder=0)
        ax.set_xlim(lims)
        ax.set_ylim(lims)
        
        # Pearson correlation
        r_val, p_val = stats.pearsonr(x[valid], y[valid])
        ax.text(
            0.05, 0.93,
            f'$r = {r_val:.2f}$',
            transform=ax.transAxes,
            fontsize=12,
            verticalalignment='top',
            bbox=dict(boxstyle='round,pad=0.25', facecolor='white', alpha=0.85, edgecolor='gray')
        )
    
    ax.set_aspect('equal', adjustable='box')
    ax.set_title(tex_labels.get(p_name, p_name), fontsize=12, pad=6)
    ax.set_xlabel('HST/NIRCam + MIRI', fontsize=12)
    ax.set_ylabel('HST/NIRCam', fontsize=12)
    ax.tick_params(labelsize=12, direction='in', top=True, right=True)
    ax.grid(True, linestyle=':', alpha=0.4)

plt.tight_layout()
out_fig = os.path.join(out_plot_dir, 'all_fit_params_miri.png')
#plt.savefig(out_fig, dpi=300, bbox_inches='tight')
print(f"Saved master parameter plot to: {out_fig}")
plt.show()
```
